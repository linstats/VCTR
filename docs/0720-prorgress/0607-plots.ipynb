```python
import numpy as np
import matplotlib.pyplot as plt

# Case 2 settings
R, S = 6, 27
r, s = 3, 14

t = np.linspace(0, 1, 500)
scale = np.sqrt(r * s / (R * S))

# Six coefficient functions
alpha1 = scale * 5 * (t - 0.2) ** 2

alpha2 = scale * (
    np.exp(-(3 * t - 1) ** 2) - 0.75
)

alpha3 = scale * np.sin(2 * np.pi * (t - 0.5))

alpha4 = scale * (
    0.45 * 5 * (t - 0.2) ** 2
    + 0.35 * (np.exp(-(3 * t - 1) ** 2) - 0.75)
    + 0.20 * np.sin(2 * np.pi * (t - 0.5))
)

alpha5 = scale * (
    1.10 * np.exp(-((t - 0.30) ** 2) / (2 * 0.08**2))
    - 0.95 * np.exp(-((t - 0.72) ** 2) / (2 * 0.11**2))
)

alpha6 = scale * (
    18 * (t - 0.2) * (t - 0.55) * (t - 0.85) - 0.05
)

functions = [alpha1, alpha2, alpha3, alpha4, alpha5, alpha6]

fig, axes = plt.subplots(
    nrows=2,
    ncols=3,
    figsize=(12, 6),
    sharex=True,
)

for k, (ax, values) in enumerate(zip(axes.flat, functions), start=1):
    ax.plot(t, values, color="tab:blue", linewidth=2)
    ax.axhline(0, color="black", linewidth=0.7, linestyle="--")
    ax.set_title(
        rf"$\alpha_{{{k},{r},{s}}}(t)$",
        fontsize=13,
    )
    ax.set_xlabel(r"$t$")
    ax.set_ylabel(r"$\alpha(t)$")
    ax.set_xlim(0, 1)
    ax.grid(alpha=0.25)

fig.suptitle(
    rf"Representative coefficient functions: $R={R}, S={S}, r={r}, s={s}$",
    fontsize=15,
)
fig.tight_layout()

plt.savefig("fig_case2_At_plot.png", dpi=600, bbox_inches="tight")
plt.show()
```
